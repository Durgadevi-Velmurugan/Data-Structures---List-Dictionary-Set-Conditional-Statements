Sets (Operations):

Create a set called my_set with following values:['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i']


```python
my_set=set(['a','e','i','o','u','a','a','i'])
print(my_set)
```

Explanation :
 Given values contain 8 elements but set only contain 5 values because set only allow unique values so duplicate 'a' and 'i' are removed during set creation.

Attempt to change the value of my_set[4] = 's'. If code throws an error, provide
an explanation.

```python
my_set[4] = 's'
```

Explanation: Set is a un-order collections and don't have fixed position like lists so set does not support indexing so we can't change the value  of "my_set" to [4]=s.
